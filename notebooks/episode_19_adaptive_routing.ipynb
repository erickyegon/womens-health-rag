```python
from rag.agent.graph import run_agent
test_cases = [
    ('Hello, what can you help me with?', 'direct'),
    ('What is the MMR in Nigeria 2021?', 'vector or hybrid'),
    ('How does female education affect child mortality in Kenya?', 'multihop'),
]
for question, expected in test_cases:
    result = run_agent(question)
    print(f'Q: {question[:65]}')
    print(f'Expected: {expected}')
    print(f'Got:      {result.get("strategy")}')
    print(f'Rewrites: {result.get("rewrites",0)}')
    print()
```
