```python
import sys,json
from pathlib import Path
cwd=Path().resolve(); repo_root=cwd.parent if cwd.name=='notebooks' else cwd
sys.path.insert(0,str(repo_root/'src'))
from dotenv import load_dotenv; load_dotenv(repo_root/'.env',override=True)
print('ready')
```

```python
# Stream the agent to see the self-correction loop
from rag.agent.graph import stream_agent
question = 'What corrections were made in the Kenya DHS erratum FR380erratum?'
print(f'Q: {question}')
print()
for node_name, update in stream_agent(question):
    print(f'→ NODE: {node_name}')
    if 'rewrites' in update:
        print(f'  rewrites so far: {update["rewrites"]}')
    if 'graded_docs' in update:
        print(f'  relevant docs:   {len(update["graded_docs"])}')
    if 'answer' in update and update['answer']:
        print(f'  answer generated: {len(update["answer"])} chars')
    if 'grounded' in update:
        print(f'  grounded:        {update["grounded"]}')
```
