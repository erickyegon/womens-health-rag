```python
# Test the API (must be running: make up)
import requests
API = 'http://localhost:8000'
HEADERS = {'Authorization': 'Bearer dev-secret-change-in-prod'}
# Health check
try:
    r = requests.get(f'{API}/health', timeout=5)
    print('Health:', r.json())
except Exception as e:
    print(f'API not running: {e}')
    print('Run: make up  to start the full stack')
```

```python
# Sync query
try:
    r = requests.post(f'{API}/query',headers=HEADERS,
                      json={'question':'What is the TFR in Nigeria?','stream':False},
                      timeout=30)
    data = r.json()
    print(f'Answer: {data["answer"][:200]}')
    print(f'Sources: {len(data["sources"])}')
    print(f'Elapsed: {data["metadata"].get("elapsed")}s')
except Exception as e:
    print(f'Error: {e}')
```
