```python
import sys,json
from pathlib import Path
cwd=Path().resolve(); repo_root=cwd.parent if cwd.name=='notebooks' else cwd
sys.path.insert(0,str(repo_root/'src'))
from dotenv import load_dotenv; load_dotenv(repo_root/'.env',override=True)
```

```python
from rag.retrieval.vector_retriever import VectorRetriever
ret = VectorRetriever()
docs = ret.retrieve('contraceptive prevalence', top_k=10)
countries = [d.metadata.get('country','?') for d in docs]
from collections import Counter
print('Countries in results (unfiltered):')
for c,n in Counter(countries).most_common(): print(f'  {c}: {n}')
```

```python
docs_filtered = ret.retrieve('contraceptive prevalence', top_k=5,
                               filters={'country':'Kenya'})
print('Countries after filter: {"Kenya" only}')
for d in docs_filtered:
    print(f'  {d.metadata.get("country")} {d.metadata.get("year")} | {d.page_content[:100]}...')
```

```python
from rag.retrieval.self_query import SelfQueryRetriever
sqr = SelfQueryRetriever()
queries = [
    'What does the 2022 Kenya report say about contraception?',
    'Nigeria DHS 2021 maternal mortality',
    'Female education trends in Ghana',
]
for q in queries:
    docs, filters = sqr.retrieve(q)
    print(f'Query: {q}')
    print(f'Extracted filters: {filters}')
    print(f'Results: {len(docs)} docs from {set(d.metadata.get("country","") for d in docs)}')
    print()
```

```python
# Compare: same question, with and without filter extraction
q = 'What is the contraceptive prevalence in Kenya 2022?'
docs_raw = ret.retrieve(q, top_k=5)
docs_sq, filters = sqr.retrieve(q)
print('WITHOUT filter extraction:')
for d in docs_raw[:3]: print(f'  {d.metadata.get("country","?")} | {d.page_content[:80]}...')
print()
print(f'WITH filter extraction (filters={filters}):')
for d in docs_sq[:3]: print(f'  {d.metadata.get("country","?")} | {d.page_content[:80]}...')
```
